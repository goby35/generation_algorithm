```python
import pandas as pd
import numpy as np
import random
import time
```

```python
file_path = 'data.xlsx'

# Đọc sheet 1 từ excel
all_course = pd.read_excel(file_path, sheet_name=0)

# Khởi tạo danh sách khóa học sau khi được xử lý
courses = []

# Duyệt qua từng dòng của DataFrame và thêm vào danh sách courses
for index, row in all_course.iterrows():
    # Chuyển đổi mỗi dòng thành dictionary và thêm vào courses
    courses.append(row.to_dict())

# In các khóa học sau khi xử lý
for course in courses:
    print(course)
```

```python
# Đọc sheet 2 từ excel
all_instructor = pd.read_excel(file_path, sheet_name=2)

#Khởi tạo danh sách room
instructors = []

# Duyệt qua từng hàng trong DataFrame và thêm vào rooms
for index, row in all_instructor.iterrows():
    instructors.append(row.to_dict())

for instructor in instructors:
    print(instructor)
```

```python
week_days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Satuday"]
time_slots = (
            [{"day": day, "session": "Morning"} for day in week_days] +
            [{"day": day, "time": "Afternoon"} for day in week_days])*2
```

```python
#Dữ liệu này dùng để test và biểu diễn trong bài báo cáo và slide thuyết trình
courses = [
    {"Code": "CS101", "Name": "Introduction to Computer Science"},
    {"Code": "MATH201", "Name": "Calculus I"},
    {"Code": "ENG301", "Name": "English Literature"},
    {"Code": "HIST101", "Name": "World History"},
    {"Code": "BIO101", "Name": "Biology Basics"},
]
instructors = [
    {'Instructor': 'Mr. Lưu Tiến Đạo'},
    {'Instructor': 'Mr. Trần Cao Đệ'},
    {'Instructor': 'Mrs. Lâm Nhựt Khang'},
    {'Instructor': 'Mrs. Trần Nguyễn Minh Thư'},
    {'Instructor': 'Mr. Ngô Bá Hùng'}
]
rooms = [
    {"room": "Room 101"},
    {"room": "Room 102"},
    {"room": "Room 201"},
    {"room": "Room 202"},
    {"room": "Room 301"},
]
```

```python
import random

# Hàm đánh giá độ phù hợp của một cá thể (schedule) dựa trên các ràng buộc
def calculate_fitness(schedule):
    penalty = 10  # Điểm khởi đầu, càng cao càng tốt. Điểm sẽ giảm khi vi phạm ràng buộc

    # Ràng buộc 1: Phòng phải trống trong thời gian đó
    used_rooms = {}
    for entry in schedule:
        # Khóa duy nhất gồm ngày, thời gian (session hoặc time), và phòng học
        key = (entry["time"]["day"], entry["time"].get("session", entry["time"].get("time")), entry["room"]["room"])
        if key in used_rooms:
            penalty -= 1  # Phạt nếu có hai lớp học trong cùng thời gian và phòng
        else:
            used_rooms[key] = True  # Đánh dấu phòng đã được sử dụng

    # Ràng buộc 2: Giảng viên có trống giờ để dạy không
    instructor_schedule = {}
    for entry in schedule:
        day = entry["time"]["day"]  # Lấy giá trị ngày học từ dictionary "time"
        session = entry["time"].get("session", entry["time"].get("time"))  # Lấy buổi học
        instructor = entry["instructor"]["Instructor"]  # Lấy tên giảng viên từ dictionary "instructor"

        # Khóa duy nhất cho giảng viên là (ngày, session)
        instructor_key = (day, session, instructor)
        if instructor_key in instructor_schedule:
            penalty -= 1  # Phạt nếu giảng viên đã có lớp trong thời điểm đó
        else:
            instructor_schedule[instructor_key] = True  # Đánh dấu giảng viên đã có lịch dạy trong thời điểm đó

    return penalty  # Trả về điểm độ phù hợp

# Hàm khởi tạo quần thể ban đầu gồm các cá thể ngẫu nhiên
def create_population(size=10):
    population = []
    for _ in range(size):
        individual = []
        for course in courses:
            schedule_entry = {
                "course": course["Code"],  # Mã lớp học
                "instructor": random.choice(instructors),  # Chọn giảng viên ngẫu nhiên
                "room": random.choice(rooms),  # Chọn phòng ngẫu nhiên
                "time": random.choice(time_slots)  # Chọn thời gian ngẫu nhiên
            }
            individual.append(schedule_entry)
        population.append(individual)  # Thêm cá thể vào quần thể
    return population

# Hàm chọn lọc các cá thể có độ phù hợp cao nhất
def selection(population):
    population.sort(key=lambda x: calculate_fitness(x), reverse=True)  # Sắp xếp cá thể theo điểm phù hợp
    return population[:len(population) // 2]  # Giữ lại 50% cá thể tốt nhất

# Hàm lai ghép (crossover) giữa hai cá thể cha mẹ
def crossover(parent1, parent2):
    crossover_point = random.randint(1, len(parent1) - 1)  # Chọn điểm lai ghép ngẫu nhiên
    child = parent1[:crossover_point] + parent2[crossover_point:]
    return child


def mutate(individual):
    index = random.randint(0, len(individual) - 1)  # Chọn vị trí ngẫu nhiên trong cá thể
    individual[index]["instructor"] = random.choice(instructors)  # Thay đổi giảng viên
    individual[index]["room"] = random.choice(rooms)  # Thay đổi phòng học
    individual[index]["time"] = random.choice(time_slots)  # Thay đổi thời gian

# Hàm giải thuật di truyền để tìm thời khóa biểu tối ưu
def genetic_algorithm(generations=200, population_size=100):
    population = create_population(population_size)  # Tạo quần thể ban đầu

    for generation in range(generations):
        population = selection(population)  # Chọn lọc các cá thể tốt nhất

        # Tạo thế hệ tiếp theo từ các cá thể đã chọn lọc
        next_generation = []
        while len(next_generation) < population_size:
            parent1 = random.choice(population)  # Chọn cha mẹ ngẫu nhiên
            parent2 = random.choice(population)  # Chọn cha mẹ ngẫu nhiên
            child = crossover(parent1, parent2)  # Lai ghép để tạo ra con
            if random.random() < 0.1:  # Xác suất đột biến là 10%
                mutate(child)
            next_generation.append(child)

        population = next_generation  # Cập nhật quần thể với thế hệ mới

        # Kiểm tra độ phù hợp tốt nhất trong quần thể
        best_individual = max(population, key=lambda x: calculate_fitness(x))
        print(f"Generation {generation} Best fitness: {calculate_fitness(best_individual)}")

    return best_individual  # Trả về cá thể tốt nhất (lịch trình tối ưu)

# Chạy giải thuật di truyền
optimal_schedule = genetic_algorithm()
print("Optimal Schedule:")
for entry in optimal_schedule:
    print(entry)
```

```python
# Lưu kết quả vào file Excel
def save_to_excel(schedule, filename="KQ.xlsx"):
    df = pd.DataFrame(schedule)
    df.to_excel(filename, index=False)
    print(f"Schedule saved to {filename}")

save_to_excel(optimal_schedule)

```

```python
# In một cặp cá thể bố mẹ và một cá thể con
def print_parents_and_child(population):
    parent1 = random.choice(population)
    parent2 = random.choice(population)
    child = crossover(parent1, parent2)

    print("Parent 1:")
    for entry in parent1:
        print(entry)

    print("\nParent 2:")
    for entry in parent2:
        print(entry)

    print("\nChild:")
    for entry in child:
        print(entry)

# Gọi hàm in
print_parents_and_child(create_population())
```

```python
# Hàm lưu kết quả vào file Excel, bao gồm 2 bố mẹ và 1 thế hệ con
def save_parents_and_child_to_excel(population):
    parent1 = random.choice(population)
    parent2 = random.choice(population)
    child = crossover(parent1, parent2)

    # Đưa dữ liệu vào DataFrame để xuất ra Excel
    data = {
        "Parent 1": [str(entry) for entry in parent1],
        "Parent 2": [str(entry) for entry in parent2],
        "Child": [str(entry) for entry in child]
    }

    df = pd.DataFrame(data)
    df.to_excel("VD.xlsx", index=False)
    print("Parents and child have been saved to 'VD.xlsx'")

save_parents_and_child_to_excel(create_population())
```
